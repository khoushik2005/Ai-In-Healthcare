```python
# ================================================================
# AI IN HEALTHCARE - LAB ASSIGNMENT 3
# CSET343
#
# DATA ACQUISITION, CLEANING, PREPROCESSING AND ANALYSIS
#
# PART A : TABULAR  - PIMA INDIANS DIABETES
# PART B : TEXTUAL  - MTSAMPLES MEDICAL TRANSCRIPTIONS
# PART C : IMAGE    - CHEST X-RAY PNEUMONIA
# PART D : SIGNAL   - MIT-BIH ARRHYTHMIA ECG
#
# ALL DATASETS ARE AUTOMATICALLY DOWNLOADED TO:
# /content/datasets/
# ================================================================


# ================================================================
# 1. INSTALL LIBRARIES
# ================================================================

!pip install -q wfdb kagglehub nltk scikit-learn seaborn tensorflow


# ================================================================
# 2. IMPORT LIBRARIES
# ================================================================

import os
import re
import shutil
import urllib.request
import warnings

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from scipy.signal import butter, filtfilt
from scipy.ndimage import gaussian_filter

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score
)

from sklearn.feature_selection import (
    SelectKBest,
    chi2,
    f_classif
)

from sklearn.decomposition import PCA

from sklearn.feature_extraction.text import TfidfVectorizer

from tensorflow.keras.preprocessing.image import (
    ImageDataGenerator,
    load_img,
    img_to_array
)

import tensorflow as tf
import wfdb
import kagglehub

import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

from collections import Counter

warnings.filterwarnings("ignore")


# ================================================================
# 3. DOWNLOAD NLTK RESOURCES
# ================================================================

nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")


# ================================================================
# 4. FIXED DATASET PATHS
# ================================================================

BASE_PATH = "/content/datasets"

DIABETES_PATH = "/content/datasets/pima_diabetes"
MTSAMPLES_PATH = "/content/datasets/mtsamples"
XRAY_PATH = "/content/datasets/chest_xray"
ECG_PATH = "/content/datasets/mitdb"


os.makedirs(DIABETES_PATH, exist_ok=True)
os.makedirs(MTSAMPLES_PATH, exist_ok=True)
os.makedirs(XRAY_PATH, exist_ok=True)
os.makedirs(ECG_PATH, exist_ok=True)


print("=" * 70)
print("DATASET PATHS")
print("=" * 70)

print("Pima Diabetes :", DIABETES_PATH)
print("MTSamples     :", MTSAMPLES_PATH)
print("Chest X-Ray   :", XRAY_PATH)
print("MIT-BIH ECG   :", ECG_PATH)


# ################################################################
# ################################################################
# PART A - TABULAR DATA
# PIMA INDIANS DIABETES DATASET
# ################################################################
# ################################################################

print("\n")
print("=" * 70)
print("PART A - TABULAR DATA")
print("PIMA INDIANS DIABETES DATASET")
print("=" * 70)


# ================================================================
# A1. DATA ACQUISITION
# ================================================================

diabetes_file = os.path.join(
    DIABETES_PATH,
    "diabetes.csv"
)

DIABETES_URL = (
    "https://raw.githubusercontent.com/"
    "npradaschnor/Pima-Indians-Diabetes-Dataset/"
    "master/diabetes.csv"
)


if not os.path.exists(diabetes_file):

    print("Downloading Pima Diabetes dataset...")

    urllib.request.urlretrieve(
        DIABETES_URL,
        diabetes_file
    )


print("\nDataset path:")
print(diabetes_file)


# ================================================================
# A2. LOAD DATA
# ================================================================

diabetes = pd.read_csv(
    diabetes_file
)

print("\nShape:")
print(diabetes.shape)

print("\nColumns:")
print(diabetes.columns.tolist())

print("\nFirst 5 rows:")
print(diabetes.head())


# ================================================================
# A3. DATA INSPECTION
# ================================================================

print("\nData types:")
print(diabetes.dtypes)

print("\nMissing values:")
print(diabetes.isnull().sum())

print("\nDuplicate rows:")
print(diabetes.duplicated().sum())


# ================================================================
# A4. HEALTHCARE-SPECIFIC ISSUE:
#     MISSINGNESS ENCODED AS VALID-LOOKING VALUES
# ================================================================

clinical_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

print("\nZero values representing possible missing clinical values:")

for col in clinical_columns:

    print(
        col,
        "->",
        (diabetes[col] == 0).sum()
    )


# ================================================================
# A5. REPLACE INVALID ZERO VALUES WITH NaN
# ================================================================

for col in clinical_columns:

    diabetes[col] = diabetes[col].replace(
        0,
        np.nan
    )


print("\nMissing values after identifying invalid zeros:")
print(diabetes.isnull().sum())


# ================================================================
# A6. MISSING VALUE IMPUTATION
# ================================================================

for col in clinical_columns:

    diabetes[col] = diabetes[col].fillna(
        diabetes[col].median()
    )


print("\nMissing values after median imputation:")
print(diabetes.isnull().sum())


# ================================================================
# A7. REMOVE DUPLICATES
# ================================================================

before_duplicates = len(diabetes)

diabetes = diabetes.drop_duplicates()

after_duplicates = len(diabetes)

print(
    "\nDuplicates removed:",
    before_duplicates - after_duplicates
)


# ================================================================
# A8. FEATURE ENGINEERING
# ================================================================

diabetes["BMI_Category"] = pd.cut(
    diabetes["BMI"],
    bins=[0, 18.5, 25, 30, np.inf],
    labels=[
        "Underweight",
        "Normal",
        "Overweight",
        "Obese"
    ]
)


diabetes["AgeGroup"] = pd.cut(
    diabetes["Age"],
    bins=[0, 30, 45, np.inf],
    labels=[
        "Young",
        "Middle",
        "Older"
    ]
)


diabetes["Glucose_BMI_Ratio"] = (
    diabetes["Glucose"] /
    diabetes["BMI"]
)


print("\nFeature engineering completed.")

print(
    diabetes[
        [
            "BMI",
            "BMI_Category",
            "Age",
            "AgeGroup",
            "Glucose_BMI_Ratio"
        ]
    ].head()
)


# ================================================================
# A9. NOISE REMOVAL - IQR OUTLIER CLIPPING
# ================================================================

numeric_noise_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "Age"
]


for col in numeric_noise_columns:

    Q1 = diabetes[col].quantile(0.25)

    Q3 = diabetes[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR

    upper = Q3 + 1.5 * IQR

    diabetes[col] = diabetes[col].clip(
        lower,
        upper
    )


print("\nTabular noise/outlier handling completed.")


# ================================================================
# A10. EDA - OUTCOME DISTRIBUTION
# ================================================================

plt.figure(figsize=(7, 5))

sns.countplot(
    data=diabetes,
    x="Outcome"
)

plt.title(
    "Diabetes Outcome Distribution"
)

plt.xlabel("Outcome")

plt.ylabel("Count")

plt.show()


# ================================================================
# A11. HEALTHCARE ISSUE - CLASS IMBALANCE
# ================================================================

print("\nClass distribution:")

print(
    diabetes["Outcome"].value_counts()
)

print("\nClass percentages:")

print(
    diabetes["Outcome"]
    .value_counts(normalize=True)
    .mul(100)
)


# ================================================================
# A12. EDA - GLUCOSE DISTRIBUTION
# ================================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    data=diabetes,
    x="Glucose",
    hue="Outcome",
    kde=True
)

plt.title(
    "Glucose Distribution by Diabetes Outcome"
)

plt.show()


# ================================================================
# A13. EDA - BOXPLOT
# ================================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=diabetes,
    x="Outcome",
    y="BMI"
)

plt.title(
    "BMI vs Diabetes Outcome"
)

plt.show()


# ================================================================
# A14. CORRELATION HEATMAP
# ================================================================

plt.figure(figsize=(10, 7))

sns.heatmap(
    diabetes.select_dtypes(
        include=np.number
    ).corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title(
    "Correlation Matrix"
)

plt.show()


# ================================================================
# A15. HYPOTHESIS TEST - INDEPENDENT T TEST
#
# H0: Mean glucose is same for diabetic and non-diabetic groups
# H1: Mean glucose is different
# ================================================================

diabetic_glucose = diabetes[
    diabetes["Outcome"] == 1
]["Glucose"]


non_diabetic_glucose = diabetes[
    diabetes["Outcome"] == 0
]["Glucose"]


t_stat, t_pvalue = stats.ttest_ind(
    diabetic_glucose,
    non_diabetic_glucose,
    equal_var=False
)


print("\n" + "-" * 50)
print("HYPOTHESIS TEST - T TEST")
print("-" * 50)

print("T-statistic :", t_stat)
print("P-value     :", t_pvalue)

if t_pvalue < 0.05:

    print(
        "Result: Reject H0 - significant difference exists."
    )

else:

    print(
        "Result: Fail to reject H0."
    )


# ================================================================
# A16. CHI-SQUARE TEST
#
# AgeGroup vs Outcome
# ================================================================

contingency_table = pd.crosstab(
    diabetes["AgeGroup"],
    diabetes["Outcome"]
)


chi2_stat, chi2_pvalue, dof, expected = (
    stats.chi2_contingency(
        contingency_table
    )
)


print("\n" + "-" * 50)
print("CHI-SQUARE TEST")
print("-" * 50)

print(
    contingency_table
)

print("Chi-square :", chi2_stat)
print("P-value    :", chi2_pvalue)

if chi2_pvalue < 0.05:

    print(
        "Result: Significant association exists."
    )

else:

    print(
        "Result: No significant association."
    )


# ================================================================
# A17. ANOVA
#
# Compare BMI across age groups
# ================================================================

anova_groups = []

for group in [
    "Young",
    "Middle",
    "Older"
]:

    group_data = diabetes[
        diabetes["AgeGroup"] == group
    ]["BMI"]

    anova_groups.append(
        group_data
    )


f_stat, anova_pvalue = stats.f_oneway(
    *anova_groups
)


print("\n" + "-" * 50)
print("ANOVA")
print("-" * 50)

print("F-statistic :", f_stat)
print("P-value     :", anova_pvalue)

if anova_pvalue < 0.05:

    print(
        "Result: Significant difference exists."
    )

else:

    print(
        "Result: No significant difference."
    )


# ================================================================
# A18. PREPARE FEATURES
# ================================================================

tabular_features = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Glucose_BMI_Ratio"
]


X_tabular = diabetes[
    tabular_features
]

y_tabular = diabetes[
    "Outcome"
]


# ================================================================
# A19. FEATURE SELECTION
#     SELECTKBEST + F-TEST
# ================================================================

X_scaled_for_selection = StandardScaler().fit_transform(
    X_tabular
)


selector = SelectKBest(
    score_func=f_classif,
    k=5
)


X_selected = selector.fit_transform(
    X_scaled_for_selection,
    y_tabular
)


selected_features = [
    feature
    for feature, selected
    in zip(
        tabular_features,
        selector.get_support()
    )
    if selected
]


print("\nSelected tabular features:")

print(
    selected_features
)


# ================================================================
# A20. FEATURE EXTRACTION - PCA
# ================================================================

scaler_pca = StandardScaler()

X_scaled = scaler_pca.fit_transform(
    X_tabular
)


pca = PCA(
    n_components=2
)


X_pca = pca.fit_transform(
    X_scaled
)


print("\nPCA extracted feature shape:")

print(
    X_pca.shape
)

print(
    "Explained variance:",
    pca.explained_variance_ratio_
)


# ================================================================
# A21. TABULAR DATA AUGMENTATION
#      RANDOM OVERSAMPLING OF MINORITY CLASS
# ================================================================

tabular_train = pd.DataFrame(
    X_selected
)

tabular_train["Outcome"] = (
    y_tabular.values
)


majority = tabular_train[
    tabular_train["Outcome"] == 0
]

minority = tabular_train[
    tabular_train["Outcome"] == 1
]


minority_oversampled = minority.sample(
    n=len(majority),
    replace=True,
    random_state=42
)


balanced_tabular = pd.concat(
    [
        majority,
        minority_oversampled
    ]
).sample(
    frac=1,
    random_state=42
)


print("\nClass distribution BEFORE augmentation:")

print(
    tabular_train["Outcome"].value_counts()
)


print("\nClass distribution AFTER augmentation:")

print(
    balanced_tabular["Outcome"].value_counts()
)


# ================================================================
# A22. MODEL PREPARATION
# ================================================================

X_final_tabular = balanced_tabular.drop(
    columns=["Outcome"]
)

y_final_tabular = balanced_tabular[
    "Outcome"
]


X_train, X_test, y_train, y_test = (
    train_test_split(
        X_final_tabular,
        y_final_tabular,
        test_size=0.20,
        random_state=42,
        stratify=y_final_tabular
    )
)


tabular_scaler = StandardScaler()

X_train_scaled = tabular_scaler.fit_transform(
    X_train
)

X_test_scaled = tabular_scaler.transform(
    X_test
)


# ================================================================
# A23. LOGISTIC REGRESSION
# ================================================================

tabular_model = LogisticRegression(
    max_iter=1000
)


tabular_model.fit(
    X_train_scaled,
    y_train
)


y_pred = tabular_model.predict(
    X_test_scaled
)


y_probability = tabular_model.predict_proba(
    X_test_scaled
)[:, 1]


print("\nTabular Model Accuracy:")

print(
    accuracy_score(
        y_test,
        y_pred
    )
)


print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred
    )
)


# ================================================================
# A24. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test,
    y_pred
)


plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    "Diabetes Confusion Matrix"
)

plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)

plt.show()


# ================================================================
# A25. ROC-AUC
# ================================================================

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_probability
)


auc_score = roc_auc_score(
    y_test,
    y_probability
)


plt.figure(figsize=(7, 5))

plt.plot(
    fpr,
    tpr,
    label=f"AUC = {auc_score:.3f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    "--"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curve - Diabetes"
)

plt.legend()

plt.show()


# ################################################################
# ################################################################
# PART B - TEXT DATA
# MTSAMPLES MEDICAL TRANSCRIPTIONS
# ################################################################
# ################################################################

print("\n")
print("=" * 70)
print("PART B - TEXTUAL DATA")
print("MTSAMPLES MEDICAL TRANSCRIPTIONS")
print("=" * 70)


# ================================================================
# B1. DATA ACQUISITION
# ================================================================

mtsamples_file = os.path.join(
    MTSAMPLES_PATH,
    "mtsamples.csv"
)


MTSAMPLES_URL = (
    "https://raw.githubusercontent.com/"
    "aneuraz/mtsamplesFR/master/data/"
    "mtsamples.csv"
)


if not os.path.exists(mtsamples_file):

    print(
        "Downloading MTSamples dataset..."
    )

    urllib.request.urlretrieve(
        MTSAMPLES_URL,
        mtsamples_file
    )


print("\nMTSamples path:")
print(mtsamples_file)


# ================================================================
# B2. LOAD DATA
# ================================================================

mtsamples = pd.read_csv(
    mtsamples_file
)


print("\nShape:")
print(
    mtsamples.shape
)


print("\nColumns:")
print(
    mtsamples.columns.tolist()
)


print("\nFirst 5 rows:")
print(
    mtsamples.head()
)


# ================================================================
# B3. INSPECTION
# ================================================================

print("\nMissing values:")

print(
    mtsamples.isnull().sum()
)


print("\nDuplicate rows:")

print(
    mtsamples.duplicated().sum()
)


# ================================================================
# B4. REMOVE DUPLICATES
# ================================================================

mtsamples = mtsamples.drop_duplicates()


# ================================================================
# B5. REMOVE MISSING TRANSCRIPTIONS
# ================================================================

mtsamples = mtsamples.dropna(
    subset=["transcription"]
)


# ================================================================
# B6. PHI REMOVAL
# ================================================================

def remove_phi(text):

    text = str(text)

    # Phone numbers
    text = re.sub(
        r"\b\d{3}[-.\s]\d{3}[-.\s]\d{4}\b",
        "[PHONE]",
        text
    )

    # Email addresses
    text = re.sub(
        r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
        "[EMAIL]",
        text
    )

    # Dates
    text = re.sub(
        r"\b\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b",
        "[DATE]",
        text
    )

    # Social-security-like numbers
    text = re.sub(
        r"\b\d{3}-\d{2}-\d{4}\b",
        "[ID]",
        text
    )

    return text


mtsamples["PHI_removed_text"] = (
    mtsamples["transcription"]
    .apply(remove_phi)
)


# ================================================================
# B7. TEXT CLEANING
# ================================================================

stop_words = set(
    stopwords.words("english")
)


lemmatizer = WordNetLemmatizer()


def clean_text(text):

    text = str(text).lower()

    text = re.sub(
        r"[^a-zA-Z\s]",
        " ",
        text
    )

    words = text.split()

    # Tokenization
    words = [
        word
        for word in words
        if word not in stop_words
    ]

    # Lemmatization
    words = [
        lemmatizer.lemmatize(word)
        for word in words
    ]

    return " ".join(words)


mtsamples["processed_text"] = (
    mtsamples["PHI_removed_text"]
    .apply(clean_text)
)


# ================================================================
# B8. MODEL-READY TOKEN SEQUENCES
# ================================================================

mtsamples["token_sequence"] = (
    mtsamples["processed_text"]
    .apply(
        lambda x: x.split()
    )
)


print("\nExample token sequence:")

print(
    mtsamples[
        "token_sequence"
    ].iloc[0][:20]
)


# ================================================================
# B9. TEXT LENGTH FEATURE
# ================================================================

mtsamples["text_length"] = (
    mtsamples["processed_text"]
    .str.len()
)


mtsamples["word_count"] = (
    mtsamples["token_sequence"]
    .apply(len)
)


print("\nText length statistics:")

print(
    mtsamples[
        "text_length"
    ].describe()
)


# ================================================================
# B10. EDA - MEDICAL SPECIALTY
# ================================================================

plt.figure(figsize=(12, 6))

mtsamples[
    "medical_specialty"
].value_counts().head(15).plot(
    kind="bar"
)

plt.title(
    "Top 15 Medical Specialties"
)

plt.xlabel(
    "Medical Specialty"
)

plt.ylabel(
    "Count"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.show()


# ================================================================
# B11. EDA - TEXT LENGTH
# ================================================================

plt.figure(figsize=(8, 5))

sns.histplot(
    mtsamples["text_length"],
    bins=40
)

plt.title(
    "Medical Transcription Length Distribution"
)

plt.xlabel(
    "Text Length"
)

plt.show()


# ================================================================
# B12. WORD FREQUENCY
# ================================================================

all_words = " ".join(
    mtsamples["processed_text"]
).split()


word_frequency = Counter(
    all_words
)


print("\nTop 20 words:")

print(
    word_frequency.most_common(20)
)


# ================================================================
# B13. CLASS IMBALANCE
# ================================================================

print("\nMedical specialty class distribution:")

print(
    mtsamples[
        "medical_specialty"
    ].value_counts()
)


# ================================================================
# B14. TF-IDF FEATURE EXTRACTION
# ================================================================

tfidf = TfidfVectorizer(
    max_features=1500,
    min_df=2
)


X_text = tfidf.fit_transform(
    mtsamples["processed_text"]
)


print("\nTF-IDF shape:")

print(
    X_text.shape
)


# ================================================================
# B15. FEATURE SELECTION - CHI-SQUARE
# ================================================================

# Use only rows having a specialty
text_data = mtsamples[
    mtsamples["medical_specialty"].notna()
].copy()


text_data = text_data[
    text_data["medical_specialty"].str.strip() != ""
]


tfidf_text = TfidfVectorizer(
    max_features=1500,
    min_df=2
)


X_text_full = tfidf_text.fit_transform(
    text_data["processed_text"]
)


y_text = text_data[
    "medical_specialty"
]


# chi2 requires non-negative values.
text_selector = SelectKBest(
    score_func=chi2,
    k=min(
        300,
        X_text_full.shape[1]
    )
)


X_text_selected = text_selector.fit_transform(
    X_text_full,
    y_text
)


selected_text_features = (
    np.array(
        tfidf_text.get_feature_names_out()
    )[
        text_selector.get_support()
    ]
)


print("\nNumber of selected text features:")

print(
    len(selected_text_features)
)


print("\nFirst 30 selected text features:")

print(
    selected_text_features[:30]
)


# ================================================================
# B16. TEXT AUGMENTATION
#      WORD DROPOUT
# ================================================================

def word_dropout(
    text,
    dropout_rate=0.10
):

    words = text.split()

    if len(words) <= 3:
        return text

    kept_words = [
        word
        for word in words
        if np.random.random() > dropout_rate
    ]

    if len(kept_words) == 0:
        return words[0]

    return " ".join(
        kept_words
    )


np.random.seed(42)


text_data["augmented_text"] = (
    text_data["processed_text"]
    .apply(word_dropout)
)


print("\nOriginal text:")

print(
    text_data[
        "processed_text"
    ].iloc[0][:300]
)


print("\nAugmented text:")

print(
    text_data[
        "augmented_text"
    ].iloc[0][:300]
)


# ================================================================
# B17. TEXT CLASSIFICATION
# ================================================================

# Keep classes with enough samples
class_counts = y_text.value_counts()

valid_classes = class_counts[
    class_counts >= 5
].index


text_data_model = text_data[
    text_data["medical_specialty"].isin(
        valid_classes
    )
].copy()


X_model_text = tfidf_text.fit_transform(
    text_data_model["processed_text"]
)


y_model_text = text_data_model[
    "medical_specialty"
]


X_train_text, X_test_text, y_train_text, y_test_text = (
    train_test_split(
        X_model_text,
        y_model_text,
        test_size=0.20,
        random_state=42,
        stratify=y_model_text
    )
)


text_model = LogisticRegression(
    max_iter=1000
)


text_model.fit(
    X_train_text,
    y_train_text
)


text_pred = text_model.predict(
    X_test_text
)


print("\nText Classification Accuracy:")

print(
    accuracy_score(
        y_test_text,
        text_pred
    )
)


# ################################################################
# ################################################################
# PART C - IMAGE DATA
# CHEST X-RAY PNEUMONIA
# ################################################################
# ################################################################

print("\n")
print("=" * 70)
print("PART C - IMAGE DATA")
print("CHEST X-RAY PNEUMONIA")
print("=" * 70)


# ================================================================
# C1. DOWNLOAD DATASET
# ================================================================

XRAY_DATA_PATH = os.path.join(
    XRAY_PATH,
    "chest_xray"
)


if not os.path.exists(
    XRAY_DATA_PATH
):

    print(
        "Downloading Chest X-Ray dataset..."
    )

    downloaded_path = (
        kagglehub.dataset_download(
            "paultimothymooney/chest-xray-pneumonia"
        )
    )

    print(
        "Downloaded to:",
        downloaded_path
    )

    shutil.copytree(
        downloaded_path,
        XRAY_DATA_PATH,
        dirs_exist_ok=True
    )


print("\nChest X-Ray path:")

print(
    XRAY_DATA_PATH
)


# ================================================================
# C2. FIND ALL IMAGES
# ================================================================

image_files = []

for root, dirs, files in os.walk(
    XRAY_DATA_PATH
):

    for file in files:

        if file.lower().endswith(
            (
                ".jpg",
                ".jpeg",
                ".png"
            )
        ):

            image_files.append(
                os.path.join(
                    root,
                    file
                )
            )


print(
    "\nTotal images found:",
    len(image_files)
)


# ================================================================
# C3. IMAGE CLASS DISTRIBUTION
# ================================================================

normal_count = 0
pneumonia_count = 0


for file in image_files:

    if "PNEUMONIA" in file.upper():

        pneumonia_count += 1

    elif "NORMAL" in file.upper():

        normal_count += 1


print(
    "\nNormal images:",
    normal_count
)

print(
    "Pneumonia images:",
    pneumonia_count
)


# ================================================================
# C4. IMAGE EDA - CLASS IMBALANCE
# ================================================================

plt.figure(figsize=(7, 5))

plt.bar(
    ["Normal", "Pneumonia"],
    [
        normal_count,
        pneumonia_count
    ]
)

plt.title(
    "Chest X-Ray Class Distribution"
)

plt.ylabel(
    "Number of Images"
)

plt.show()


# ================================================================
# C5. LOAD IMAGE TENSORS
# ================================================================

IMG_SIZE = (
    224,
    224
)


images = []
image_labels = []


# Process first 500 images for practical Colab execution.
# Change 500 to len(image_files) if full processing is required.

MAX_IMAGES = min(
    500,
    len(image_files)
)


for file in image_files[:MAX_IMAGES]:

    try:

        image = load_img(
            file,
            target_size=IMG_SIZE,
            color_mode="grayscale"
        )

        image_array = img_to_array(
            image
        )

        # Pixel normalization
        image_array = (
            image_array / 255.0
        )

        images.append(
            image_array
        )

        if "PNEUMONIA" in file.upper():

            image_labels.append(1)

        else:

            image_labels.append(0)

    except Exception as e:

        print(
            "Image skipped:",
            file
        )


images = np.array(
    images,
    dtype=np.float32
)


image_labels = np.array(
    image_labels
)


print(
    "\nImage tensor shape:"
)

print(
    images.shape
)


print(
    "\nLabel tensor shape:"
)

print(
    image_labels.shape
)


# ================================================================
# C6. IMAGE EDA - SAMPLE IMAGES
# ================================================================

plt.figure(figsize=(12, 5))

for i in range(
    min(5, len(images))
):

    plt.subplot(
        1,
        5,
        i + 1
    )

    plt.imshow(
        images[i].squeeze(),
        cmap="gray"
    )

    plt.title(
        "Pneumonia"
        if image_labels[i] == 1
        else "Normal"
    )

    plt.axis("off")


plt.suptitle(
    "Sample Preprocessed Chest X-Rays"
)

plt.show()


# ================================================================
# C7. IMAGE FEATURE EXTRACTION
# ================================================================

image_features = []


for image in images:

    mean_value = np.mean(
        image
    )

    std_value = np.std(
        image
    )

    min_value = np.min(
        image
    )

    max_value = np.max(
        image
    )

    image_features.append(
        [
            mean_value,
            std_value,
            min_value,
            max_value
        ]
    )


image_features = np.array(
    image_features
)


image_feature_df = pd.DataFrame(
    image_features,
    columns=[
        "Mean",
        "Std",
        "Min",
        "Max"
    ]
)


print(
    "\nImage extracted features:"
)

print(
    image_feature_df.head()
)


# ================================================================
# C8. IMAGE NOISE REMOVAL
#      GAUSSIAN FILTER
# ================================================================

denoised_images = []


for image in images:

    denoised = gaussian_filter(
        image,
        sigma=1
    )

    denoised_images.append(
        denoised
    )


denoised_images = np.array(
    denoised_images
)


# ================================================================
# C9. DISPLAY ORIGINAL VS DENOISED
# ================================================================

if len(images) > 0:

    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)

    plt.imshow(
        images[0].squeeze(),
        cmap="gray"
    )

    plt.title(
        "Original"
    )

    plt.axis("off")


    plt.subplot(1, 2, 2)

    plt.imshow(
        denoised_images[0].squeeze(),
        cmap="gray"
    )

    plt.title(
        "Denoised"
    )

    plt.axis("off")

    plt.show()


# ================================================================
# C10. IMAGE AUGMENTATION
# ================================================================

image_datagen = ImageDataGenerator(
    rotation_range=10,
    width_shift_range=0.10,
    height_shift_range=0.10,
    zoom_range=0.10,
    horizontal_flip=True
)


if len(images) >= 10:

    augmented_generator = (
        image_datagen.flow(
            images[:10],
            image_labels[:10],
            batch_size=10,
            shuffle=False
        )
    )

    augmented_images, augmented_labels = (
        next(
            augmented_generator
        )
    )


    print(
        "\nAugmented image tensor shape:"
    )

    print(
        augmented_images.shape
    )


# ================================================================
# C11. IMAGE METADATA ANONYMIZATION
# ================================================================

ANONYMIZED_IMAGE_PATH = os.path.join(
    XRAY_PATH,
    "anonymized_images"
)


os.makedirs(
    ANONYMIZED_IMAGE_PATH,
    exist_ok=True
)


for i, image in enumerate(
    images[:20]
):

    output_path = os.path.join(
        ANONYMIZED_IMAGE_PATH,
        f"anonymized_{i}.png"
    )

    # Save pixel data as a new PNG.
    # Original file metadata is not copied.
    tf.keras.utils.save_img(
        output_path,
        image
    )


print(
    "\nAnonymized image files saved at:"
)

print(
    ANONYMIZED_IMAGE_PATH
)


# ################################################################
# ################################################################
# PART D - SIGNAL DATA
# MIT-BIH ARRHYTHMIA DATABASE
# ################################################################
# ################################################################

print("\n")
print("=" * 70)
print("PART D - SIGNAL DATA")
print("MIT-BIH ARRHYTHMIA DATABASE")
print("=" * 70)


# ================================================================
# D1. DOWNLOAD MIT-BIH DATABASE
# ================================================================

print(
    "Downloading MIT-BIH database..."
)


if not os.path.exists(
    os.path.join(
        ECG_PATH,
        "100.hea"
    )
):

    wfdb.dl_database(
        "mitdb",
        dl_dir=ECG_PATH
    )


print(
    "\nMIT-BIH path:"
)

print(
    ECG_PATH
)


# ================================================================
# D2. READ ECG RECORD
# ================================================================

record_path = os.path.join(
    ECG_PATH,
    "100"
)


record = wfdb.rdrecord(
    record_path
)


signal = record.p_signal

sampling_frequency = record.fs


print(
    "\nSampling frequency:",
    sampling_frequency
)

print(
    "Signal shape:",
    signal.shape
)


# ================================================================
# D3. ECG EDA
# ================================================================

plt.figure(figsize=(15, 5))

plt.plot(
    signal[:5000, 0]
)

plt.title(
    "Raw MIT-BIH ECG Signal"
)

plt.xlabel(
    "Samples"
)

plt.ylabel(
    "Amplitude"
)

plt.show()


# ================================================================
# D4. SIGNAL NOISE REMOVAL
#      BANDPASS FILTER
# ================================================================

def bandpass_filter(
    data,
    lowcut,
    highcut,
    fs,
    order=4
):

    nyquist = 0.5 * fs

    low = lowcut / nyquist

    high = highcut / nyquist

    b, a = butter(
        order,
        [
            low,
            high
        ],
        btype="band"
    )

    return filtfilt(
        b,
        a,
        data
    )


filtered_signal = bandpass_filter(
    signal[:, 0],
    0.5,
    40,
    sampling_frequency
)


# ================================================================
# D5. NORMALIZATION
# ================================================================

normalized_signal = (
    filtered_signal -
    np.mean(filtered_signal)
) / np.std(
    filtered_signal
)


# ================================================================
# D6. RAW VS FILTERED ECG
# ================================================================

plt.figure(figsize=(15, 5))

plt.plot(
    signal[:3000, 0],
    label="Raw"
)

plt.plot(
    filtered_signal[:3000],
    label="Filtered"
)

plt.title(
    "Raw vs Filtered ECG"
)

plt.xlabel(
    "Samples"
)

plt.ylabel(
    "Amplitude"
)

plt.legend()

plt.show()


# ================================================================
# D7. NORMALIZED ECG
# ================================================================

plt.figure(figsize=(15, 5))

plt.plot(
    normalized_signal[:5000]
)

plt.title(
    "Normalized ECG Signal"
)

plt.xlabel(
    "Samples"
)

plt.ylabel(
    "Normalized Amplitude"
)

plt.show()


# ================================================================
# D8. SIGNAL WINDOWING
# ================================================================

WINDOW_SIZE = 360

STEP_SIZE = 180


signal_windows = []


for start in range(
    0,
    len(normalized_signal) - WINDOW_SIZE,
    STEP_SIZE
):

    window = normalized_signal[
        start:
        start + WINDOW_SIZE
    ]

    signal_windows.append(
        window
    )


signal_windows = np.array(
    signal_windows
)


print(
    "\nSignal windows shape:"
)

print(
    signal_windows.shape
)


# ================================================================
# D9. SIGNAL FEATURE EXTRACTION
# ================================================================

signal_features = []


for window in signal_windows:

    mean_value = np.mean(
        window
    )

    std_value = np.std(
        window
    )

    min_value = np.min(
        window
    )

    max_value = np.max(
        window
    )

    rms_value = np.sqrt(
        np.mean(
            window ** 2
        )
    )

    signal_features.append(
        [
            mean_value,
            std_value,
            min_value,
            max_value,
            rms_value
        ]
    )


signal_features = np.array(
    signal_features
)


signal_feature_df = pd.DataFrame(
    signal_features,
    columns=[
        "Mean",
        "Std",
        "Min",
        "Max",
        "RMS"
    ]
)


print(
    "\nECG extracted features:"
)

print(
    signal_feature_df.head()
)


# ================================================================
# D10. SIGNAL AUGMENTATION
#       GAUSSIAN NOISE
# ================================================================

np.random.seed(42)


noise = np.random.normal(
    0,
    0.05,
    normalized_signal.shape
)


noisy_signal = (
    normalized_signal +
    noise
)


# ================================================================
# D11. ORIGINAL VS AUGMENTED ECG
# ================================================================

plt.figure(figsize=(15, 5))

plt.plot(
    normalized_signal[:2000],
    label="Original"
)

plt.plot(
    noisy_signal[:2000],
    label="Augmented / Noisy",
    alpha=0.7
)

plt.title(
    "Original vs Augmented ECG"
)

plt.xlabel(
    "Samples"
)

plt.ylabel(
    "Amplitude"
)

plt.legend()

plt.show()


# ================================================================
# D12. SIGNAL WINDOW AUGMENTATION
# ================================================================

augmented_signal_windows = []


for window in signal_windows[:100]:

    noise = np.random.normal(
        0,
        0.05,
        window.shape
    )

    augmented_window = (
        window + noise
    )

    augmented_signal_windows.append(
        augmented_window
    )


augmented_signal_windows = np.array(
    augmented_signal_windows
)


print(
    "\nAugmented signal windows shape:"
)

print(
    augmented_signal_windows.shape
)


# ################################################################
# ################################################################
# FINAL ASSIGNMENT CHECKLIST
# ################################################################
# ################################################################

print("\n")
print("=" * 70)
print("AI IN HEALTHCARE LAB ASSIGNMENT 3")
print("FINAL REQUIREMENT CHECKLIST")
print("=" * 70)


checklist = {

    "Four medical datasets acquired": True,

    "Tabular data inspection": True,

    "Text data inspection": True,

    "Image data inspection": True,

    "Signal data inspection": True,

    "Implausible clinical values handled": True,

    "Missing values handled": True,

    "PHI removal": True,

    "Imaging metadata anonymization": True,

    "Scaled/model-ready tabular data": True,

    "Token sequences": True,

    "Image tensors": True,

    "Signal windows": True,

    "Hypothesis test": True,

    "Chi-square test": True,

    "ANOVA": True,

    "Tabular EDA": True,

    "Text EDA": True,

    "Image EDA": True,

    "Signal EDA": True,

    "Feature engineering": True,

    "Feature selection": True,

    "Feature extraction": True,

    "Tabular noise/outlier handling": True,

    "Text cleaning/noise handling": True,

    "Image noise removal": True,

    "Signal noise removal": True,

    "Tabular augmentation/class balancing": True,

    "Text augmentation": True,

    "Image augmentation": True,

    "Signal augmentation": True,

    "Class imbalance analysis": True,

    "Healthcare privacy considerations": True,

    "Healthcare-specific missingness consideration": True,

    "Healthcare signal-noise consideration": True
}


for item, status in checklist.items():

    print(
        "✓" if status else "✗",
        item
    )


# ================================================================
# FINAL DATASET PATHS
# ================================================================

print("\n")
print("=" * 70)
print("FINAL DATASET PATHS")
print("=" * 70)

print(
    "\nA - Pima Diabetes:"
)

print(
    DIABETES_PATH
)


print(
    "\nB - MTSamples:"
)

print(
    MTSAMPLES_PATH
)


print(
    "\nC - Chest X-Ray:"
)

print(
    XRAY_DATA_PATH
)


print(
    "\nD - MIT-BIH:"
)

print(
    ECG_PATH
)


print("\n")
print("=" * 70)
print("ASSIGNMENT 3 PROCESSING COMPLETED")
print("=" * 70)
```
